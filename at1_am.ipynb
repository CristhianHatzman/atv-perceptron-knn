# Desafio 1 - Perceptron com Treinamento Automático

Neste primeiro desafio será implementado um Perceptron para classificar transações como legítimas ou suspeitas.

Os atributos considerados são

- Valor normalizado da transação;
- Frequência de operações recentes.

As classes são

- `0` - Transação legítima;
- `1` - Transação suspeita.

## Dados de treinamento

Os dados abaixo representam transações já conhecidas.

Cada linha possui duas características:

1. Valor normalizado da transação;
2. Frequência de operações recentes.

As classes representam:

- `0`: transação legítima;
- `1`: transação suspeita.

```python
# X representa as características das transações.
# Cada linha é uma transação e cada coluna representa uma característica.
X_treino_fraude = np.array([
    [1.5, 1.0],
    [2.0, 2.0],
    [3.5, 1.5],
    [3.0, 3.0],
    [6.5, 5.0],
    [7.0, 7.0],
    [8.5, 6.0],
    [9.0, 8.0]
])

# y contém a classificação correta de cada transação.
# 0 = legítima
# 1 = suspeita
y_treino_fraude = np.array([
    0, 0, 0, 0,
    1, 1, 1, 1
])

print("Dados de entrada:")
print(X_treino_fraude)

print("\nClasses:")
print(y_treino_fraude)


def funcao_degrau(valor):
    # Se o valor calculado pelo Perceptron for maior ou igual a zero,
    # a amostra será classificada como classe 1.
    if valor >= 0:
        return 1

    # Caso contrário, será classificada como classe 0.
    return 0


def treinar_perceptron(X, y, taxa_aprendizado=0.1, epocas=100):

    pesos = np.zeros(X.shape[1])

    # O viés também começa em zero.
    vies = 0.0

    # O treinamento poderá repetir no máximo 100 vezes.
    # Cada repetição completa recebe o nome de época.
    for epoca in range(epocas):

        erros = 0

        # Percorre simultaneamente as entradas e suas classes corretas.
        for entrada, classe_real in zip(X, y):
           
            soma = np.dot(entrada, pesos) + vies

            # Transforma o resultado em classe 0 ou 1.
            classe_predita = funcao_degrau(soma)

            # Calcula a diferença entre o resultado correto
            # e a previsão realizada pelo Perceptron.
            erro = classe_real - classe_predita

            # Só é necessário alterar os pesos se houver erro.
            if erro != 0:

                # Atualização dos pesos utilizando a regra
                # de aprendizagem do Perceptron.
                pesos += taxa_aprendizado * erro * entrada

                # Atualização do viés.
                vies += taxa_aprendizado * erro

                # Registra que ocorreu um erro nesta época.
                erros += 1

        # Se nenhuma classificação estiver errada,
        # significa que o modelo conseguiu aprender os dados.
        if erros == 0:
            print(f"Treinamento concluído na época {epoca + 1}")
            break

    # Retorna os valores aprendidos pelo modelo.
    return pesos, vies
```

```python
pesos, vies = treinar_perceptron(
    X_treino_fraude,
    y_treino_fraude
)

# Mostra os valores encontrados pelo algoritmo.
print("\nPesos finais:", pesos)
print(f"Viés final: {vies:.2f}")


# Após o treinamento, os pesos e o viés já foram aprendidos.
# A função abaixo utiliza esses valores para classificar novas transações que não participaram do treinamento.

def prever_perceptron(amostra, pesos, vies):

    # Aplica os pesos aprendidos à nova amostra.
    soma = np.dot(amostra, pesos) + vies

    # A função degrau transforma o resultado em classe 0 ou 1.
    return funcao_degrau(soma)


# Antes de testar novas transações, verificamos se o Perceptron consegue classificar corretamente os próprios dados utilizados durante o treinamento.
# Contador de previsões corretas.
acertos = 0

print("Verificação das amostras de treinamento:\n")

# Percorre todas as amostras utilizadas durante o treinamento.
for entrada, classe_real in zip(X_treino_fraude, y_treino_fraude):

    # Faz uma previsão utilizando os pesos aprendidos.
    classe_predita = prever_perceptron(
        entrada,
        pesos,
        vies
    )

    print(
        f"Entrada: {entrada} | "
        f"Esperado: {classe_real} | "
        f"Previsto: {classe_predita}"
    )

    # Verifica se a previsão está correta.
    if classe_predita == classe_real:
        acertos += 1


# Calcula a porcentagem de acertos.
acuracia = acertos / len(y_treino_fraude) * 100

print(f"\nAcurácia no treinamento: {acuracia:.2f}%")

# Agora o modelo será utilizado para analisar duas transações que não participaram diretamente do treinamento.

transacao_A = np.array([2.5, 2.0])
transacao_B = np.array([8.0, 6.5])

resultado_A = prever_perceptron(
    transacao_A,
    pesos,
    vies
)

resultado_B = prever_perceptron(
    transacao_B,
    pesos,
    vies
)

# Converte o resultado numérico para uma descrição mais fácil de interpretar.
print(
    "Transação A:",
    "Suspeita" if resultado_A == 1 else "Legítima"
)

print(
    "Transação B:",
    "Suspeita" if resultado_B == 1 else "Legítima"
)
```

# Desafio 2 - Classificador K-Nearest Neighbors (KNN)

Neste desafio será implementado um classificador KNN para identificar o risco de clientes.

Cada cliente possui duas características

- Dias de inatividade;
- Quantidade de chamados críticos.

As classes utilizadas são

- `0` - Baixo Risco;
- `1` - Alto Risco.

O algoritmo irá procurar os `K` clientes mais próximos de uma nova entrada e utilizar a classe predominante entre eles para realizar a classificação.

```python
# Dados utilizados como referência pelo KNN.
# Cada linha representa um cliente.
#
# Coluna 1 = dias de inatividade
# Coluna 2 = chamados críticos

X_treino_churn = np.array([
    [2.0, 1.0],
    [3.0, 0.0],
    [5.0, 1.0],
    [6.0, 2.0],
    [15.0, 4.0],
    [18.0, 5.0],
    [20.0, 4.0],
    [22.0, 6.0]
])


y_treino_churn = np.array([
    0, 0, 0, 0,
    1, 1, 1, 1
])

print("Dados de treinamento:")
print(X_treino_churn)

print("\nClasses:")
print(y_treino_churn)
```

```python
def classificar_knn(X, y, ponto, k=3, metrica="euclidiana"):

    # Garante que o ponto recebido esteja no formato de array NumPy.
    ponto = np.array(ponto)

    # Calcula a distância entre o novo ponto e todas as amostras.
    if metrica == "euclidiana":

        # Distância Euclidiana:
        # raiz quadrada da soma das diferenças ao quadrado.
        distancias = np.sqrt(
            np.sum((X - ponto) ** 2, axis=1)
        )

    elif metrica == "manhattan":

        # Distância Manhattan:
        # soma das diferenças absolutas.
        distancias = np.sum(
            np.abs(X - ponto),
            axis=1
        )

    else:
        raise ValueError(
            "Métrica inválida. Utilize 'euclidiana' ou 'manhattan'."
        )

    # Ordena as distâncias e seleciona os índices
    # dos K vizinhos mais próximos.
    indices_vizinhos = np.argsort(distancias)[:k]

    # Recupera as classes desses vizinhos.
    classes_vizinhos = y[indices_vizinhos]

    # Conta quantas vezes cada classe aparece.
    contagem_classes = np.bincount(classes_vizinhos)

    # A classe com maior quantidade de votos será escolhida.
    classe_predita = np.argmax(contagem_classes)

    return (
        classe_predita,
        indices_vizinhos,
        distancias[indices_vizinhos]
    )
```

## Clientes utilizados para teste

Serão classificados dois novos clientes:

- Cliente 1: `[4.0, 1.0]`
- Cliente 2: `[17.0, 5.0]`

Para a classificação serão considerados os três vizinhos mais próximos:

`K = 3`

```python
cliente_1 = np.array([4.0, 1.0])
cliente_2 = np.array([17.0, 5.0])

K = 3
```

## Classificação utilizando distância Euclidiana

A distância Euclidiana representa a distância em linha reta entre dois pontos.

```python
classe_1_euclidiana, vizinhos_1_euclidiana, distancias_1_euclidiana = (
    classificar_knn(
        X_treino_churn,
        y_treino_churn,
        cliente_1,
        k=K,
        metrica="euclidiana"
    )
)

classe_2_euclidiana, vizinhos_2_euclidiana, distancias_2_euclidiana = (
    classificar_knn(
        X_treino_churn,
        y_treino_churn,
        cliente_2,
        k=K,
        metrica="euclidiana"
    )
)
```

```python
nomes_classes = {
    0: "Baixo Risco",
    1: "Alto Risco"
}

print("=== DISTÂNCIA EUCLIDIANA ===")

print("\nCliente 1")
print("Classificação:", nomes_classes[classe_1_euclidiana])
print("Índices dos vizinhos:", vizinhos_1_euclidiana)
print("Distâncias:", distancias_1_euclidiana)

print("\nCliente 2")
print("Classificação:", nomes_classes[classe_2_euclidiana])
print("Índices dos vizinhos:", vizinhos_2_euclidiana)
print("Distâncias:", distancias_2_euclidiana)
```

## Classificação utilizando distância Manhattan

A distância Manhattan calcula a distância somando as diferenças absolutas entre cada característica.

```python
classe_1_manhattan, vizinhos_1_manhattan, distancias_1_manhattan = (
    classificar_knn(
        X_treino_churn,
        y_treino_churn,
        cliente_1,
        k=K,
        metrica="manhattan"
    )
)

classe_2_manhattan, vizinhos_2_manhattan, distancias_2_manhattan = (
    classificar_knn(
        X_treino_churn,
        y_treino_churn,
        cliente_2,
        k=K,
        metrica="manhattan"
    )
)
```

```python
print("=== DISTÂNCIA MANHATTAN ===")

print("\nCliente 1")
print("Classificação:", nomes_classes[classe_1_manhattan])
print("Índices dos vizinhos:", vizinhos_1_manhattan)
print("Distâncias:", distancias_1_manhattan)

print("\nCliente 2")
print("Classificação:", nomes_classes[classe_2_manhattan])
print("Índices dos vizinhos:", vizinhos_2_manhattan)
print("Distâncias:", distancias_2_manhattan)
```

# Desafio 3 - Recomendador por Similaridade Espacial

Neste desafio será implementado um sistema simples de recomendação de servidores.

Cada servidor possui três características

- Quantidade de vCPUs;
- Memória RAM em GB;
- Armazenamento SSD em GB.

O objetivo é comparar o perfil desejado pelo cliente com os servidores disponíveis e recomendar os servidores mais próximos desse perfil.

Neste caso, serão retornadas as duas melhores opções.

```python
# Catálogo de servidores disponíveis.
# Cada linha representa um servidor.

catalogo_servidores = np.array([
    [2.0, 4.0, 50.0],
    [4.0, 8.0, 100.0],
    [8.0, 16.0, 250.0],
    [16.0, 32.0, 500.0],
    [32.0, 64.0, 1000.0],q
    [64.0, 128.0, 2000.0]
])

# Nomes correspondentes a cada servidor.
nomes_servidores = [
    "Micro Instância Web",
    "Standard App Server",
    "Medium Backend & Cache",
    "Database Enterprise",
    "High Performance Computing",
    "Big Data & AI Training"
]

print("Catálogo de servidores:")

for i, servidor in enumerate(catalogo_servidores):
    print(
        f"{i} - {nomes_servidores[i]} | "
        f"{servidor[0]:.0f} vCPU | "
        f"{servidor[1]:.0f} GB RAM | "
        f"{servidor[2]:.0f} GB SSD"
    )
```

## Perfil desejado

O cliente procura um servidor com as seguintes características:

- 12 vCPUs;
- 28 GB de memória RAM;
- 450 GB de armazenamento SSD.

O sistema irá calcular a distância entre esse perfil e todos os servidores disponíveis.

```python
# Perfil solicitado pelo cliente.
perfil_desejado = np.array([
    12.0,
    28.0,
    450.0
])

print("Perfil desejado:")
print("vCPUs:", perfil_desejado[0])
print("RAM:", perfil_desejado[1], "GB")
print("SSD:", perfil_desejado[2], "GB")
```

## Funcionamento do recomendador

O sistema utiliza a distância Euclidiana para medir o quanto cada servidor está próximo do perfil desejado.

Quanto menor for a distância, mais parecido o servidor é com a configuração solicitada.

O processo funciona da seguinte forma:

1. Calcula a distância entre o perfil desejado e todos os servidores;
2. Ordena as distâncias da menor para a maior;
3. Seleciona os dois servidores mais próximos;
4. Exibe um ranking com as melhores opções.


```python
def recomendar_servidores(catalogo, perfil, k=2):

    # Calcula a distância Euclidiana entre o perfil desejado
    # e todos os servidores do catálogo.
    distancias = np.sqrt(
        np.sum((catalogo - perfil) ** 2, axis=1)
    )

    # Ordena os servidores da menor distância para a maior
    # e seleciona apenas os K mais próximos.
    indices = np.argsort(distancias)[:k]

    # Retorna os índices dos servidores escolhidos
    # e suas respectivas distâncias.
    return indices, distancias[indices]
```

## Geração das recomendações

Serão selecionados os dois servidores mais próximos do perfil solicitado.

Para isso:

`K = 2`

```python
K = 2

indices_recomendados, distancias_recomendadas = recomendar_servidores(
    catalogo_servidores,
    perfil_desejado,
    k=K
)
```

```python
print("=== RECOMENDAÇÃO DE SERVIDORES ===")

for posicao, (indice, distancia) in enumerate(
    zip(indices_recomendados, distancias_recomendadas),
    start=1
):

    servidor = catalogo_servidores[indice]

    print(f"\n{posicao}º lugar")
    print("Servidor:", nomes_servidores[indice])
    print(f"vCPUs: {servidor[0]:.0f}")
    print(f"RAM: {servidor[1]:.0f} GB")
    print(f"SSD: {servidor[2]:.0f} GB")
    print(f"Distância: {distancia:.2f}")
```

## Análise dos resultados

O sistema comparou o perfil desejado com todos os servidores disponíveis utilizando a distância Euclidiana.

O servidor `Database Enterprise` foi a opção mais próxima do perfil solicitado e, por isso, ficou em primeiro lugar no ranking.

O servidor `Medium Backend & Cache` apareceu como a segunda opção mais próxima.

Neste desafio não existe uma classe como no Perceptron ou no KNN. O objetivo é apenas encontrar os elementos que possuem características mais semelhantes ao perfil desejado.

Quanto menor for a distância calculada, maior é a similaridade entre o servidor e a necessidade do cliente.
